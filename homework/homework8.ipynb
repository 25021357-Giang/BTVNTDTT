W8A16 Yêu cầu

Xây dựng lớp Shape với phương thức area().

Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().

Lấy π = 3.14.

Input

Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)

Dòng 2: tham số tương ứng.

Output

In ra diện tích với 2 chữ số thập phân.

```python
class Shape:
    def area(self):
        return 0.0

class Rectangle(Shape):
    def __init__(self,w,h):
        self.w=w
        self.h=h
    
    def area(self):
        return self.w*self.h

class Circle(Shape):
    def __init__(self,r):
        self.r=r

    def area(self):
        return 3.14*(self.r**2)

def main():
    n=input().strip().upper()

    if n=="RECTANGLE":
        w,h=map(int,input().split())

        shape=Rectangle(w,h)
    elif n=="CIRCLE":
        r=float(input().strip())

        shape=Circle(r)
    else:
        return 
    print(f"{shape.area():.2f}")
if __name__=="__main__":
    main()
    
```

Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức
add_interest() để cộng balance*rate vào số dư.
Input
balance rate
Output
Số dư sau khi cộng lãi, in 2 chữ số thập phân. 

```python
class Account:
    def __init__(self,balance):
        self.balance=balance

    def deposit(self,x):
        self.balance+=x

    def get_balance(self):
        return self.balance
    
class SavingAccount(Account):
    def __init__(self,balance,rate):
        super().__init__(balance)
        self.rate=rate

    def add_interest(self):
        self.balance+=self.balance*self.rate
        return self.balance
    
balance,rate=map(float,input().split())
lai=SavingAccount(balance,rate)
print(f"{lai.add_interest():.2f}")

        
```

W8A25 Yêu cầu
Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất:
P(x) = ax + b, Q(x) = cx + d.
Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.
Input
a b c d
Output
Chuỗi của đa thức P(x)+Q(x). 


```python
class PolynomialSum:
    def __init__(self, a: float, b: float, c: float, d: float):
        self.k = a + c
        self.m = b + d

    def sum_str(self) -> str:
        if self.k == 0 and self.m == 0:
            return "0"
        
        parts = []
        
        if self.k != 0:
            if self.k == 1:
                parts.append("x")
            elif self.k == -1:
                parts.append("-x")
            else:
                parts.append(f"{self.k}x")
        
        if self.m != 0:
            if self.m > 0 and self.k != 0:
                parts.append(f"+ {self.m}")
            elif self.m > 0 and self.k == 0:
                parts.append(f"{self.m}")
            else:  
                if self.k != 0:
                    parts.append(f"- {abs(self.m)}")
                else:
                    parts.append(f"{self.m}")
                    
        return " ".join(parts).replace(" +", " +").replace(" -", " -")


if __name__ == "__main__":

    a, b, c, d = map(float, input().split())
    
    
    poly_sum = PolynomialSum(a, b, c, d)
    print(poly_sum.sum_str())
```

W8A27 Yêu cầu
Xây dựng lớp Product(name, price).
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức
final_price() để trả về giá sau giảm.
Công thức: price * (100 - percent) / 100
Input
name price percent
Output
In ra giá cuối cùng với 2 chữ số thập phân. 

```python
class Product:
    def __init__(self,name,price):
        self.name=name
        self.price=float(price)
    
    def final_price(self):
        return self.price

class DiscountProduct(Product):
    def __init__(self,name,price,percent):
        super().__init__(name,price)
        self.percent=float(percent)

    def final_price(self):
        return f"{self.price*(100-self.percent)/100:.2f}"
    
name,price,percent=input().split()
gia=DiscountProduct(name,price,percent)
print(gia.final_price())
        
```

W8A28 Yêu cầu
Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(),
value().
Sau đó thực hiện dãy thao tác trên đối tượng.
Input
Dòng 1: n
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.
Output
Giá trị cuối cùng của count

```python
class Counter:
    def __init__(self):
        self.count=0

    def inc(self):
        self.count+=1

    def dec(self):
        self.count-=1

    def value(self):
        return self.count
    
n=int(input())
thaotac=input().split()
tong=Counter()
for i in thaotac:
    i=i.upper()
    if i=="INC":
        tong.inc()
    elif i=="DEC":
        tong.dec()
print(tong.value())
```
