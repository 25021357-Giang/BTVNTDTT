W9A2
Cho ma trận hai chiều nxm có các phần tử được sắp xếp trên
dòng tăng dần và trên cột tăng dần, hãy tìm số lớn thứ k (k <
nxm) với thuật toán tốt nhất có thể

```python
matrix = []

while True:
    s = input()
    if len(s.split()) == 1:
        k = int(s)
        break

    matrix.append(list(map(int, s.split())))


def count_ge(matrix, x):
    n = len(matrix)
    m = len(matrix[0])

    i = n - 1
    j = 0
    cnt = 0

    while i >= 0 and j < m:
        if matrix[i][j] >= x:
            cnt += (m - j)
            i -= 1
        else:
            j += 1

    return cnt


def kth_largest(matrix, k):
    low = matrix[0][0]
    high = matrix[-1][-1]

    ans = low

    while low <= high:
        mid = (low + high) // 2

        if count_ge(matrix, mid) >= k:
            ans = mid
            low = mid + 1
        else:
            high = mid - 1

    return ans


print(kth_largest(matrix, k))
```

W9A3Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ
số để được một số nguyên nhỏ nhất có thể 


```python
a=input().strip()
def doi(a):
    lst=list(a)
    n=len(a)
    
    for i in range(n):
        min=i
        for j in range(n-1,i,-1):
            if lst[min]>lst[j]:
                min=j
            
        if min!= i:
            lst[i],lst[min]=lst[min],lst[i]
            return "".join(lst)
    lst[-1],lst[-2]=lst[-2],lst[-1]
    return "".join(lst)
print(int(doi(a)))
```

W9A5 Cho hai mảng số nguyên list1 và list2, viết hàm kiểm tra xem list1
có thuộc list2 hay không (có phải là dãy con liên tiếp của list2 hay
không)? 


```python
def kt(lst1,lst2):
    if len(lst1)>len(lst2):
        return "NO"

    a=lst1[0]
    b=len(lst1)
    cnt=0
    
    for i in range(len(lst2)):
        if lst2[i]==a:
            for j in range(b):
                if lst1[j]==lst2[i+j]:
                    cnt+=1
                else:
                    return "NO"
            if cnt==b:
                return "YES"
            else:
                cnt=0
                
            
lst1=list(map(int,input().split()))
lst2=list(map(int,input().split()))
print(kt(lst1,lst2))
```
