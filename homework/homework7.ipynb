Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):


```python
def binary_search(arr,left,right,target):
    while left<=right:
        mid=(left+right)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]<target:
            left=mid+1
        else:
            right = mid-1
    return -1
arr=list(map(int,input().split()))
left,right=0,len(arr)-1
x=int(input())
print(binary_search(arr,left,right,x))
```

Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list

```python
intervals = eval(input()) 
queries = eval(input())
res = []
for q in queries:
  ans = float('inf')
  for l, r in intervals:
    if l<=q<=r:
      ans = min(ans, r-l+1)
  if ans == float("inf"):
    res.append (-1)
  else:
    res.append(ans)
print (res)
```

W7A9
Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree

```python
def find_insertion_index(arr, target):
    left, right = 0, len(arr) - 1
    idx = len(arr)  

    while left <= right:
        mid = (left + right) // 2
        if arr[mid] >= target:
            idx = mid  
            right = mid - 1  
        else:
            left = mid + 1 

    return idx


def count_smaller(nums):
    seen = []  
    result = [0] * len(nums)

    for i in range(len(nums) - 1, -1, -1):
        count = find_insertion_index(seen, nums[i])
        result[i] = count
        seen.insert(count, nums[i])

    return result

nums = list(map(int, input("Nhập mảng số: ").split()))
print("Kết quả:", count_smaller(nums))
```

Tìm phần tử xuất hiện nhiều hơn n/2 lần
trong mảng (nếu có) với độ phức tạp thời
gian là O(n), độ phức tạp bộ nhớ là O(1),
- Boyer–Moore. Nếu không có phần tử
nào thỏa mãn thì không in gì


```python
def dem(n):
    cnt=0
    can=None
    for i in n:
        if cnt==0:
            can=i
            cnt=1
        elif can==i:
            cnt+=1
        else:
            cnt-=1
    count=0
    for i in n:
        if can==i:
            count+=1
    if count>(len(n)//2):
        return can
lst=list(map(int,input().split()))
print(dem(lst))
```

W7A25
Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số

```python
def tang(n:str,k:int)->str:
    n=n.strip()
    l=len(n)
    h=l-k
    ds=[]
    if k >= l:
        return n
    for i in n:
        while h>0 and ds and ds[-1]<i:
            ds.pop()
            h-=1
        ds.append(i)
    return "".join(ds[:k])
n=input()
k=int(input())
print(tang(n,k))
```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra
xem dãy ngoặc đã cho có hợp lệ hay
không

```python
def isValid(s: str) -> bool:    
    stack = []
    mapping = {")": "(", "}": "{", "]": "[",">":"<"}

    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'

            if mapping[char] != top_element:
                return False
        else:

            stack.append(char)


    return not stack


test_string = input().strip()
print(isValid(test_string))  
```

```python
n=list(map(int,input().split()))
tong=[]
def day(i):
    ds=[n[i]]
    for a in range(i,len(n)-1):
        if n[a]<n[a+1]:
            ds.append(n[a+1])
        else:
            return ds
    return ds    
for i in range(len(n)):
    tong.append(day(i))
if tong:
    to=max(tong,key=len)
    print(*to)
else:
    print("end=")
```
